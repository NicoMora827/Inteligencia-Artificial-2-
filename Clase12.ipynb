Árboles de decisión: modelos de aprendizaje automático que toman decisiones mediante preguntas sucesivas sobre los datos. Cada división crea ramas hasta llegar a una predicción.

Entrenando árboles: el algoritmo busca las preguntas o condiciones que mejor separan los datos. El proceso continúa hasta cumplir ciertos criterios de parada.

Predicciones y probabilidades: en clasificación, una hoja predice una clase y puede indicar la probabilidad de pertenecer a cada clase. En regresión, la hoja predice normalmente un valor numérico.

CART e impureza: CART (Classification and Regression Trees) construye árboles mediante divisiones binarias. Para clasificación suele utilizar la impureza de Gini, que mide qué tan mezcladas están las clases dentro de un nodo. Menor impureza significa grupos más homogéneos.

Regularización: evita que el árbol memorice los datos (sobreajuste). Se puede controlar su profundidad, número mínimo de muestras por hoja, número máximo de hojas, entre otros parámetros.

Árboles de regresión: se utilizan cuando la variable objetivo es numérica. Cada hoja suele representar el promedio de los valores de entrenamiento que llegan a ella.

Limitaciones: pueden sobreajustar fácilmente, ser sensibles a pequeños cambios en los datos y producir modelos menos precisos que métodos como Random Forest o Gradient Boosting en algunos problemas. Además, los árboles individuales pueden ser inestables.
