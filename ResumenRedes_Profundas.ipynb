**ENTRENANDO REDES PROFUNDAS**

Entrenar una red neuronal profunda consiste en enseñarle a aprender patrones a partir de datos. Para lograrlo, la red ajusta sus pesos durante el entrenamiento con el objetivo de reducir los errores y mejorar sus predicciones.

**Reto de los Gradientes**

Los gradientes indican cómo deben modificarse los pesos de una red neuronal durante el entrenamiento. En redes profundas pueden aparecer problemas como el desvanecimiento o la explosión de los gradientes, lo que dificulta el aprendizaje.

**Inicialización**

Es el proceso de asignar valores iniciales a los pesos de la red. Una buena inicialización ayuda a que el entrenamiento sea estable y evita problemas con los gradientes. Algunos métodos conocidos son Xavier y He.

**Funciones de Activación**

Permiten que la red aprenda relaciones complejas entre los datos. Algunas funciones comunes son ReLU, Sigmoid y Tanh. ReLU es una de las más utilizadas en redes profundas.
Normalización
Consiste en mantener los valores que pasan por la red dentro de rangos adecuados. La normalización ayuda a estabilizar y acelerar el entrenamiento. Un ejemplo es Batch Normalization.

**Transferencia**

El aprendizaje por transferencia consiste en utilizar una red que ya fue entrenada con muchos datos y adaptarla para realizar una nueva tarea. Esto permite ahorrar tiempo y utilizar menos datos para entrenar el nuevo modelo.

**Preentrenamiento**

Consiste en entrenar previamente una red con una gran cantidad de datos para que aprenda características generales. Después, este conocimiento puede utilizarse para resolver una tarea específica.
