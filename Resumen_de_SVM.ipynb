Los Vectores de Soporte (SVM) son un modelo de Machine Learning que se utiliza principalmente para clasificar datos, aunque también puede utilizarse para regresión y detección de datos. Su objetivo es encontrar una frontera que permita separar las diferentes clases de datos de la mejor manera posible.

Una de las características principales de SVM es el margen amplio. El modelo busca una frontera que tenga la mayor distancia posible respecto a los datos más cercanos. Estos datos que se encuentran en los límites son llamados vectores de soporte, porque son los que ayudan a definir la posición de la frontera.

Cuando los datos no pueden separarse mediante una línea recta, SVM puede utilizar métodos para trabajar con datos no lineales. Una de estas técnicas es el mapeo polinomial, que transforma las características originales agregando nuevas dimensiones para que los datos puedan ser separados mediante un modelo lineal.

También se explica el truco del kernel, que permite obtener el efecto de transformar los datos a dimensiones superiores sin tener que realizar directamente toda esa transformación. Esto hace que SVM pueda trabajar de manera más eficiente con problemas no lineales.

Otro concepto importante es el margen blando, que permite que algunos datos puedan quedar dentro del margen o incluso clasificarse incorrectamente. El comportamiento se controla mediante el hiperparámetro C: un valor bajo permite más errores y un margen más amplio, mientras que un valor alto busca reducir los errores de clasificación.

SVM también puede utilizarse para regresión. En este caso, en lugar de buscar una frontera que separe clases, intenta encontrar una zona alrededor de la predicción donde los errores estén dentro de un valor llamado épsilon (ε).

Finalmente, el video muestra que SVM puede utilizar diferentes kernels y niveles de regularización para adaptarse a problemas lineales y no lineales. La elección del modelo depende de la cantidad de datos, la complejidad del problema y el tipo de relación que exista entre las características.
